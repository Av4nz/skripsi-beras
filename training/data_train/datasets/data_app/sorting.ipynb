```python
import pandas as pd
```

```python
# Load the dataset
df = pd.read_csv('datasets/komoditas_beras_2022_2026.csv')

# Display the first few rows to verify
print("Initial Data:")
display(df.head())
```

```python
# Filter by Province_ID
target_province_id = 15

filtered_df = df[df['Province_ID'] == target_province_id]

# Display the filtered data
print(f"\nData for Province ID {target_province_id}:")
display(filtered_df)
```

```python
# Export the filtered data to a CSV file
output_filename = f'filtered_data_province_yogyakarta.csv'
filtered_df.to_csv(output_filename, index=False)
print(f"Data exported to {output_filename}")
```
